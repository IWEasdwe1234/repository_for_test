## G-02_조업이벤트.csv

### EVT_ID (이벤트 번호)
약어풀이: <u>**Ev**</u>en<u>**t**</u> <u>**ID**</u>entifier\
의미: 이벤트 번호 (이벤트를 식별하는 고유 ID)

---

### EVT_DT (발생 시각)
약어풀이: <u>**Ev**</u>en<u>**t**</u> <u>**D**</u>ate <u>**T**</u>ime\
의미: 발생 시각 (이벤트가 발생한 일시)

---

### EVT_TYPE (이벤트 유형)
약어풀이: <u>**Ev**</u>en<u>**t**</u> <u>**Type**</u>\
의미: 이벤트 유형 (강종, 원료배합, 조업률 변경)

| 이벤트 유형 | 공장 코드 |
|---|---|
| <u>**강종변경**</u> | <u>**`CR1`**</u> |
| 원료배합변경 | `BF1`, `BF2` |
| 조업률변경 | `SP1`, `SP2` |

---

### PLANT_CD (공장 코드)
약어풀이: <u>**Plant**</u> <u>**C**</u>o<u>**d**</u>e\
의미: 공장 코드 (이벤트가 발생한 설비/공장 식별 코드)

| 공장 코드 | 약어 풀이 |이벤트 유형 |
|---|---|---|
| <u>**`CR1`**</u> | <u>**C**</u>old <u>**R**</u>olling (냉연 공장 / Cold Rolling Mill) | <u>**강종변경**</u> |
| `BF1`, `BF2` | <u>**B**</u>last <u>**F**</u>urnace (고로 / 용광로) | 원료배합변경 |
| `SP1`, `SP2` | <u>**S**</u>teelmaking <u>**P**</u>lant (제강 공장) | 조업률변경 |

---
